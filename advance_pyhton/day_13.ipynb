 TOPIC 1 — List Comprehensions

```python
# normal way
squares1 = []
for i in range(10):
    squares1.append(i**2)

# list comprehension
squares2 = [i**2 for i in range(10)]
print(squares1)
print(squares2)

# Only even numbers
evens = [i for i in range(20) if i % 2 == 0]
print(evens)

# transform strings
names = ["tony", "pepper", "rhodey"]
upper_names = [name.upper() for name in names]
print(upper_names)
```

* When to use it: Any time you're building a new list by looping through something and optionally filtering. If your loop is just appending to a list — use a list comprehension instead.
* When NOT to use it: When the logic inside is complex — more than one condition or nested operations. Readability matters more than cleverness.

TOPIC 2 — enumerate()

```python
# normal way
subjects = ["English","Maths","History","Physics"]
count = 0

for subject in subjects:
    count += 1
    print(f"{count}. {subject}")

# using enumerate
for count, subject in enumerate(subjects, start=1):
    print(f"{count}. {subject}")
```

enumerate() returns two things on each iteration the index and the value. start=1 means start counting from 1 instead of 0.

Without start it defaults to 0:

```python
for i, name in enumerate(["Tony", "Pepper", "Rhodey"]):
    print(i, name)
```

TOPIC 3 — zip()

zip() takes two or more lists and pairs them up element by element, like a zipper

```python
names = ["Tony", "Pepper", "Rhodey"]
scores = [95, 88, 79]

for name, score in zip(names, scores):
    print(f"{name}: {score}")

# You can zip more than two lists
names = ["Tony", "Pepper"]
scores = [95, 88]
grades = ["A", "B"]

for name, score, grade in zip(names, scores, grades):
    print(f"{name}: {score} ({grade})")

# Creating a dictionary from two lists
keys = ["name", "age", "city"]
values = ["Tony", 19, "Mumbai"]

person = dict(zip(keys, values))
print(person)

# Important behavior, if lists have different lengths, zip stops at the shortest one
a = [1, 2, 3, 4, 5]
b = [10, 20, 30]

list(zip(a, b))
# [(1, 10), (2, 20), (3, 30)] — stops at 3
```

TOPIC 4 — *args and **kwargs

*args — allows a function to accept any number of positional arguments:

The * collects all positional arguments into a tuple called args. You can name it anything *numbers, *values.. but *args is the convention.

```python
def add_all(*args):
    return sum(args)

print(add_all(1, 2, 3))        # 6
print(add_all(1, 2, 3, 4, 5))  # 15
```

**kwargs — allows a function to accept any number of keyword arguments

The ** collects all keyword arguments into a dictionary called kwargs

** unpacks a dictionary into keyword arguments. So instead of writing each parameter manually, you unpack the whole dictionary. This is why it's used when looping through multiple configurations — clean and concise.

```python
def my_function(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} = {value}")

my_function(name="Alice", age=30)
```

TOPIC 5 — Exception Handling

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
except ValueError as e:
    print(f"Value error: {e}")
except Exception as e:
    print(f"Something went wrong: {e}")
else:
    print("No error occurred")      # runs only if no exception
finally:
    print("This always runs")       # runs no matter what

'''
Multiple exceptions in one line

except (ValueError, TypeError):
    print("Wrong type or value")
'''
```

* except Exception as e — catches any exception and stores the error message in e. Use this as a catch-all when you don't know exactly what might go wrong.
* else — runs only when no exception occurred. Cleaner than putting success code inside try.
* finally — always runs, exception or not. Used for cleanup — closing files, closing database connections.

```python
# Raising your own exceptions
def get_grade(cgpa):
    if cgpa < 0 or cgpa > 10:
        raise ValueError(f"CGPA must be between 0 and 10, got {cgpa}")
    if cgpa >= 9:
        return "Outstanding"

try:
    print(get_grade(11))
except ValueError as e:
    print(e)   # CGPA must be between 0 and 10, got 11

# raise lets you create your own errors with custom messages, 
# This is how professional code communicates problems clearly
```

Task 4 — Exception Handling (20 mins)

```python
# Build a safe calculator function that:
 
# Takes two numbers and an operator
# Raises ValueError if operator is not +, -, *, /
# Raises ZeroDivisionError with custom message if dividing by zero
# Raises TypeError if inputs are not numbers
# Has a finally block that prints "Calculation attempted"
# Test it with at least 4 different inputs including error cases


def calculator(x, y, op):
    try:
        if not isinstance(x, (int, float)) or not isinstance(y, (int, float)):
            raise TypeError("Inputs must be numbers")
        
        if op not in ["+", "-", "*", "/"]:
            raise ValueError("Invalid operator")
        
        if op == "/" and y == 0:
            raise ZeroDivisionError("Cannot divide by zero")
        
        if op == "+": return x + y
        elif op == "-": return x - y
        elif op == "*": return x * y
        elif op == "/": return x / y

    except ZeroDivisionError as e:
        print(e)
    except ValueError as e:
        print(e)
    except TypeError as e:
        print(e)
    finally:
        print("Calculation attempted")

c1 = calculator(3,4,"+")
c2 = calculator(12,76,"*") 
c3 = calculator(12,10,"-") 
c4 = calculator(4,2,"/") 
c5 = calculator(4,0,"/")
c6 = calculator(4,0,"x") 


print(c1)
print(c2)
print(c3)
print(c4)
print(c5)
print(c6)

```
